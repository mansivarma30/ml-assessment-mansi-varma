# Q2: Unsupervised Learning — Customer Segmentation

This notebook performs customer segmentation using **K-Means clustering** and visualises results with **PCA**.  
Dataset columns: `age, annual_spend, visits_per_month, basket_size, days_since_last_visit, num_categories_purchased`.


```python
# Step 1: Data Preparation (3 marks)
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv('../data/q2_customers.csv')

print("Shape:", df.shape)
print("Missing values:\n", df.isnull().sum())
df.head()

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

### Step 1 Interpretation
- The dataset contains customer demographics and shopping behaviour.  
- Scaling is essential before applying K-Means because clustering relies on distance calculations.  
- Without scaling, features with larger numeric ranges (e.g., `annual_spend`) would dominate the clustering results, leading to biased clusters.


```python
# Step 2: Choosing K — Elbow Method (5 marks)
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.show()
```

### Step 2 Interpretation
- The elbow plot shows the Within-Cluster Sum of Squares (WCSS) for K = 1 to 10.  
- The optimal K is chosen at the "elbow point," where adding more clusters does not significantly reduce WCSS.  
- For example, if the elbow is at K=4, we select 4 clusters as the optimal segmentation.

```python
# Step 3: K-Means Clustering (6 marks)
optimal_k = 4  # replace with your chosen K from elbow method
kmeans = KMeans(n_clusters=optimal_k, random_state=42)
clusters = kmeans.fit_predict(X_scaled)

df['cluster'] = clusters

# Cluster centroids
centroids = pd.DataFrame(kmeans.cluster_centers_, columns=df.columns[:-1])
centroids
```

### Step 3 Interpretation
- Each customer has been assigned to a cluster.  
- The centroids represent the average profile of each cluster.  
- Example interpretations:  
  - **Cluster 0**: Young, frequent low-spenders.  
  - **Cluster 1**: Older, high-spending customers with large basket sizes.  
  - **Cluster 2**: Moderate spenders with regular visits.  
  - **Cluster 3**: Infrequent visitors but very high spenders.  
- These insights help businesses tailor marketing strategies to different customer groups.

```python
# Step 4: Dimensionality Reduction with PCA (5 marks)
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Explained Variance Ratio:", pca.explained_variance_ratio_)

loadings = pd.DataFrame(pca.components_, columns=df.columns[:-1], index=['PC1', 'PC2'])
loadings
```

### Step 4 Interpretation
- PCA reduces the dataset to 2 principal components for visualisation.  
- The explained variance ratio shows how much information each component captures.  
- Feature loadings indicate which variables contribute most:  
  - **PC1** may capture spending behaviour (annual_spend, basket_size).  
  - **PC2** may capture visit frequency and recency (visits_per_month, days_since_last_visit).  
- This helps us understand the main behavioural dimensions driving customer segmentation.


```python
# Step 5: Cluster Visualisation (3 marks)
plt.figure(figsize=(8,6))
plt.scatter(X_pca[:,0], X_pca[:,1], c=df['cluster'], cmap='viridis', s=50)
plt.title("Customer Segmentation (PCA Projection)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.colorbar(label='Cluster')
plt.show()
```

### Step 5 Interpretation
- The scatter plot shows customers projected onto two principal components, coloured by cluster.  
- Clear separation between clusters indicates meaningful segmentation.  
- Overlaps suggest customers with similar behaviours across groups.  
- This visualisation helps validate the clustering results and provides intuitive insights for business decision-making.
